```python
#Pandas 
import numpy as np
import pandas as pd
data=np.array([[15,125,45,85],[12,172,81,105],[29,161,59,115],[18,179,75,120]])
print(data.ndim)
print(type(data))
```

```python
df=pd.DataFrame(data)
df
```

```python
#giving columns name 
df.columns=["Age","Height","Weight","Bp"]
df
```

```python
#giving row names
df.index=["a","b","c","d"]
df
```

```python
#Calculations on particular column
print(df["Bp"].min())
print(df["Bp"].max())
print(df["Bp"].mean())
print(df["Bp"].median())
print(df["Bp"].mode())
print(df["Bp"].std())
print(df["Bp"].var())
```

```python
#Describe fucntion gives  the statistical calculations
df.describe()
```

```python
#calling by column names
print(df)
print(df[["Age","Height"]])
print(df[["Age","Weight"]])
```

```python
#two function for slicing
#iloc based on index values 
#loc  based on names

#iloc
print("--------iloc------")
print(df.iloc[0:2,1:3])
print(df.iloc[1:2,0:2])

print("------loc--------")
print(df.loc[['b','c'],["Age","Bp"]])
df.loc[['a','d'],["Height","Weight"]]


```
