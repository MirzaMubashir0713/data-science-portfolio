```python
#5 BOX PLOT
'''
-->it is used to find the outliers
-->first draw box plot and find outliers
-->calculte Q1,Q3,IQR,lower,upper
-->remove or replace using clip
'''

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df=pd.read_csv("market.csv")
df
```

```python
#MATPLOTLIB VERSION BOX PLOT OF SALES
plt.figure(figsize=(10,5))
plt.boxplot(df["Sales"])
plt.title("Box plot for sales(matplotlib)")
plt.ylabel(" Sales ")
plt.show()
```

```python
#SEABORN VERSION
plt.figure(figsize=(10,5))
sns.boxplot(df["Sales"])
plt.title("box plot for sales (Seaborn)")
plt.show()
```

```python
#Calculating outliers
Q1=df["Sales"].quantile(0.25)
Q2=df["Sales"].quantile(0.50)
Q3=df["Sales"].quantile(0.75)
IQR=Q3-Q1
lower=Q1-1.5*IQR
upper=Q3+1.5*IQR

print("Q1",Q1)
print("Q2",Q2)
print("Q3",Q3)
print("IQR",IQR)
print("lower",lower)
print("upper",upper)
```

```python
#Remove outliers
df=df[(df["Sales"]>=lower) & (df["Sales"]<=upper)]
#df = df[(df["Sales"] >= lower) & (df["Sales"] <= upper)]
plt.boxplot(df["Sales"])
```

```python
sns.boxplot(df["Sales"])
```

```python
df["Sales"]=df["Sales"].clip(lower,upper)
sns.boxplot(df["Sales"])
```

```python
#INVENTORY BOX PLOT
plt.figure(figsize=(10,5))
plt.boxplot(df["Inventory"])
plt.title("Box plot of inventory")
plt.ylabel("inventory")
plt.show()
```

```python
#SEABORN VERSION
plt.figure(figsize=(10,5))
sns.boxplot(df["Inventory"])
plt.title("box plot of inventory (seaborn)")
plt.show()
```

```python
#By product checking 
plt.figure(figsize=(10,5))
sns.boxplot(data=df,x="Product",y="Inventory")
plt.title("inventory distribution on product")
plt.xticks(rotation=45)
plt.show()
```

```python
Q1=df["Inventory"].quantile(0.25)
Q3=df["Inventory"].quantile(0.75)
IQR=Q3-Q1
lower=Q1-1.5*IQR
upper=Q3+1.5*IQR
```

```python
df=df[(df["Inventory"]>=lower) & (df["Inventory"]<=upper)]
df
```

```python
df["Inventory"]=df["Inventory"].clip(lower,upper)
```
