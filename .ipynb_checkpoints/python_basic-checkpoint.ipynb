```python
#Collection data types
'''
list 
tuple
set 
dictionary
'''

#List
lst1=[2,5,7,3,4,1]
print(lst1)
print(len(lst1))
#dir(lst1) #shows the list methods

#LIST METHODS
"""
 'append',
 'clear',
 'copy',
 'count',
 'extend',
 'index',
 'insert',
 'pop',
 'remove',
 'reverse',
 'sort'
"""
#ADDING ELEMENTS-----------------------------
list=[7,1,2,9,6,15,11,14]
list.reverse() #it will reverse the string
print(list)
list.append(2) #add value at end
print(list)
list.extend([2,15,45])  #add multiple value at end
print(list)
list.insert(3,17) #take the index vlaue where we want to add the value
print(list)
print("actual list after updating ==",list)
#REMOVING ELEMENTS----------------------------
list.pop() #remove the last value
print(list)
list.remove(17) #as per our request removes the value
print(list)
list.sort() #sort the values into ascending order
print(list)
print(list.index(7)) #provides the index position of particular value
print(list.count(2)) #count() is used to count how many times an element appears in a list

#slicing on list
print(list[2:5])
print(list[:9])
print(list[4:])
print(list[:])
print(list[2:15:2])


list.clear() #clear all the list
print(list)
name=['M','u','b','a','s','h','i','r']
name.insert(0,"M , i , r , z , a")
print(name)
name.reverse() 
print(name)
name.sort()
print(name)
```

```python
#COPY METHODS
#1.Shallow copy (Two different memory addres and different objects )
my_list=["apple","mango","banana"]
print(my_list)
new_list=my_list.copy()
print(new_list)
print(id(my_list))
print(id(new_list))

#2.Deep copy  (having the same memory addres)
s=['a','b','c']
ss=s
print(s)
print(ss)
print(id(s))
print(id(ss))
```

```python
#Tuple 
names=("aman","ayan","farzan","kamran")
print(names)
print(type(names))

#Slicing on tuple
print(names[1:3])
print(names[:2])
print(names[2:])
print(names[:])
print(names[0:3:2])
```

```python
#Set
'''doesnt allow duplicates and dont allow to perform indexing and slicing'''
obj={1,2,3,1,4,3}
print(obj)
print(type(obj))
```

```python
#Dictonary 
'''stores in key value pair
   allows insertions and removing updating'''

Emp={
    "name":"Mubashir",
    "age":22,
    "dob":"05-02-2004",
    "salary":"50000",
    "Designation" : "jr Data Scientist"
}
print(Emp)
print(type(Emp))

Emp["exp"] = "2 years"
print(Emp)
print(Emp["name"])
Emp.pop("age")
print(Emp)
Emp.clear()
print(Emp)
```
