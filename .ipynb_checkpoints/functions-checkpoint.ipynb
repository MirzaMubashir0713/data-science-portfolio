```python
#Functions
# fucntions are the reusable code 
''' functions are two types 
1. nomral fucntion (print,len,id,type)
2. userdefine fucntion (nomral and anonymous fucntion)

Function Terminology:
1. def function
2. fucntion name 
3. parameters
4. doc string
5. fucntion body 
6. return keyword
7. fucntion calling 
8. arguments
'''

def students(a,b,c,d):#pass #using pass keyword in this can prevent from error 
    print("first student",a)
    print("second student",b)
    print("third student",c)
    print("fourth student",d)

students("aman","ayan","farzan","kamran")

print("--------add two numbers----------------")
def add(a,b):
    c=a+b
    return c
print(add(2,10))

print("--------even or odd-----------")
def checking(e,o):
    even=e%2==0
    odd=o%2!=0
    return even,odd
print(checking(7,3))


def find(n):
    if n%2==0:
        return "even"
    else:
        return "odd"
print(find(2))
```
