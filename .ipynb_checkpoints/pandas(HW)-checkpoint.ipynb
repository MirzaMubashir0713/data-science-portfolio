```python
import numpy as np
import pandas as pd
df=pd.DataFrame(
    {
        'emp_id':[101,102,103,104,105,105],
        'name'  : ["raj","john","priya","amit","sara","sara"],
        'department':["IT","HR","IT","FINANCE","HR","HR"],
        'salary' : [5000,np.nan,6000,5500,np.nan,np.nan],
        'age'    : [25,30,28,35,32,32]
    }
)
#information of data
df.info()
```

```python
df[["name"]]
```

```python
df[["name","age"]]
```

```python
df.drop("emp_id",axis=1)
```

```python
#Average salary by Department
avg_salary=int(df.groupby("department")["salary"].mean())
print(avg_salary)
```

```python
mean=int(df["salary"].mean())
df["salary"].fillna(value=mean,inplace=True)
```
