```python
#Linear Regression
import pandas as pd
df = pd.read_csv("experience_vs_salary.csv")
df
```

```python
X=df[["Experience (Years)"]]    # Input varaible 
Y=df["Salary ($)"]              # Targetted variable
from sklean.linear_model import LinearRegression         # Skit learn is a library that provides ML Algorithms or models
model=LinearRegression()        # Select the model
model.fit(X,Y)                  # Fit the model
y_pred=model.predict(X)
print(y_pred)
```
