```python
#2. Scatter Plot 
# --> it is used to find the relation between the two columns and creates the visualization
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df=pd.read_csv("market.csv")
df
```

```python
#MATPLOTLIB VERSION FOR SALES AND INVENTORY
plt.figure(figsize=(8,5))
plt.scatter(df["Sales"],df["Inventory"])
plt.title("sales vs inventory using matplotlib")
plt.xlabel("sales")
plt.ylabel("inventory")
plt.show()
```

```python
#SEABORN VERSION FOR SALES AND INVENTORY
plt.figure(figsize=(8,5))
sns.scatterplot(data=df,x="Sales",y="Inventory")
plt.title("sales vs inventory usign seaborn")
plt.show()
```

```python
#MATPLOTLIB VERSION FOR SALES VS RETURN 
plt.figure(figsize=(8,5))
plt.scatter(df["Sales"],df["Returns"])
plt.title("sales vs returns using matplotlib")
plt.xlabel("sales")
plt.ylabel("inventory")
plt.show()
```

```python
#SEABORN VERSION FOR SALES AND RETURNS
plt.figure(figsize=(8,5))
sns.scatterplot(data=df,x="Sales",y="Returns")
plt.title("sales vs returns using seaborn")
plt.show()
```
