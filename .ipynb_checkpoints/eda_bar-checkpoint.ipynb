```python
#BAR GRAPH
#--> it is used on catogery data
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df=pd.read_csv("market.csv")
df
```

```python
#MATPLOTLIB VERSION FOR AVG SALES BY PRODUCT
avg_sales=df.groupby("Product")["Sales"].mean()
plt.figure(figsize=(10,8))
avg_sales.plot(kind="bar")
plt.title("average sales on product ( Matplotlib )")
plt.xlabel("Products")
plt.ylabel("Sales")
plt.xticks(rotation=45)  #it will rorates the names of xlabel
plt.show()
```

```python
#SEABORN VERSION FOR AVG SALES BY PRODUCT
plt.figure(figsize=(10,5))
sns.barplot(data=df,x="Product",y="Sales")
plt.title("Average sales on product (Seaborn)")
plt.xticks(rotation=45)
plt.show()
```

```python
#MATPLOTLIB VERSION AVG RETURNS ON PRODUCT
avg_returns=df.groupby("Product")["Returns"].mean()
plt.figure(figsize=(10,5))
avg_returns.plot(kind="bar")
plt.title("Average returns on products (Matplotlib)")
plt.xlabel("Products")
plt.ylabel("Returns")
plt.xticks(rotation=45)
plt.show()
```

```python
#SEABORN VERSION FOR AVG RETURNS ON PRODUCTS
plt.figure(figsize=(10,5))
sns.barplot(data=df,x="Product",y="Returns")
plt.title("Avgerage returns on products (Seaborn)")
plt.xticks(rotation=45)
plt.xlabel("Products")
plt.ylabel("Returns")
plt.show()
```
