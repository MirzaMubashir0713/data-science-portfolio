```python
X=df[["Experience (Years)"]]    # Input varaible 
Y=df["Salary ($)"]              # Targetted variable
from sklearn.linear_model import LinearRegression         # Skit learn is a library that provides ML Algorithms or models
model=LinearRegression()        # Select the model
model.fit(X,Y)                  # Fit the model
y_pred=model.predict(X)
print(y_pred)
```

```python
#Find error now
'''
   1. MAE (Mean Absolute Error)
   2. MSE (Mean Square Error)
   3. RMSE (Root Mean Square Error)
   4. R-2 Score 
---> if R-2 value is near to 1 then the model is good
'''
```

```python
from sklearn.metrics import mean_squared_error,r2_score
mse=mean_squared_error(Y,y_pred)
r2=r2_score(Y,y_pred)
print("Mean Squared Error : ", mse)
print("R2 Score : " , r2)

```

```python
#Multi Colinearity
'''
  --> when we have high corelation btw two indipendent variable 
  ---> it affects model performace
  ---> To avoid we need to find VIF (varaince Inflation Factor)

  VIF = 1 / (1 - r2_score)
  
'''
print()
```

```python
# VIF of Hp and Sp 
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error , r2_score
X_sp=df[["SP"]]
Y_hp=df["HP"]
model_sp_hp=LinearRegression()
model_sp_hp.fit(X_sp,Y_hp)
y_hp_pred=model_sp_hp.predict(X_sp)
r2_sp_hp=r2_score(Y_hp,y_hp_pred)
VIF_sp_hp=1/(1-r2_sp_hp)
print(VIF_sp_hp)
```

```python
# VIF of Sp and Vol
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error , r2_score
X_sp=df[["SP"]]
Y_vol=df["VOL"]
model_sp_vol=LinearRegression()
model_sp_vol.fit(X_sp,Y_vol)
y_vol_pred=model_sp_vol.predict(X_sp)
r2_sp_vol=r2_score(Y_vol,y_vol_pred)
VIF_sp_vol=1/(1-r2_sp_vol)
print(VIF_sp_vol)
```
