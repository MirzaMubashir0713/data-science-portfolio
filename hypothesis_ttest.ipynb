```python
#Inferential Statistics
'''
---> Inferential statistics is a branch of statistics that uses sample data to make predictions , conclusions on entire population data
---> for checking this conclusion or prediction is true or not we uses " HYPOTHESIS TESTING "
'''

#code
import numpy as np
import pandas as pd
sample = pd.read_csv("Bulbs_Lifespan.csv")
```

```python
sample.info()
```

```python
from scipy.stats import ttest_1samp

#population mean 
pop_mean=1000

#perfrom one sample test
t_statistic , p_value = ttest_1samp(sample,pop_mean)

#significance value
alpha=0.05

#results
print("t_statistic : " , np.round(t_statistic,4))
print("p_value : ", np.round(p_value,4))
```

```python
if p_value < alpha :
    print("reject the null hypothesis : the mean of bulb life span is significantly different from 1000")
else :
    print("fails to reject the null hypothesis : the mean of bulb lifespan is not significantly different from 1000")
```

```python
#Alternative Hypothesis 
#two tail  H1:u!=1000
#left tail H1:u<1000

if t_statistic < 0:
    one_tail_p_value=p_value/2
    if one_tail_p_value < alpha :
        print("reject to null hypothesis : the mean life span is significantly less than 1000 ")
    else :
        print("fails to reject null hypothesis : no significance evidence that the mean of bulb life span is less than 1000  ")
else : 
    print("fails to reject : test statistic does not support H1:u<1000 ")
    

    

```

```python
#Alternative Hypothesis 
#right tail H1:u>1000

if t_statistic > 0:
    one_tail_p_value=p_value/2
    if one_tail_p_value < alpha :
        print("reject to null hypothesis : the mean life span is significantly greater than 1000 ")
    else :
        print("fails to reject null hypothesis : no significance evidence that the mean of bulb life span is greater than 1000  ")
else : 
    print("fails to reject : test statistic does not support H1:u>1000 ")
```
