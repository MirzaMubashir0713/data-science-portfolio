```python
#ztest
import numpy as np
import pandas as pd
df=pd.read_csv("ecommerce_two_sample_ztest_dataset.csv")
df
```

```python
'''
An e-commerce company wants to compare the average purchase value of customers who visited their website through 
email marketing campaigns (Group 1) versus those who visited through social media ads (Group 2). The goal is to 
determine whether there is a significant difference in the average purchase value between the two groups
'''
from statsmodels.stats.weightstats import ztest

email_group=df[df["Group"]=="Email Marketing"]["Purchase_Value"]
social_group=df[df["Group"]=="Social Media Ads"]["Purchase_Value"]
print(email_group)
print(social_group)
```

```python
email_group.describe()
```

```python
social_group.describe()
```

```python
#perfoming two sample test
z_stats , p_value = ztest(email_group,social_group,alternative="two-sided")

#Significance value
alpha = 0.05

#results
print(z_stats)
print(p_value)
```

```python
if p_value < alpha :
    conclusion = "reject the null hypothesis : there is significance different in average puchase value"
else:
    conclusion = "fail to reject the null hypothesis : there is no significance difference in average purchase value"

#Output results
print(f"Z_statistic : {z_stats:.2f}")
print(f"P_value : {p_value:.2f}")
print(f"Conclusions : {conclusion}")
```
