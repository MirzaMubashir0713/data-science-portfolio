```python
#PIE CHART 
#--> it is used when we have less number of catogery 
#--> it is only available for matplotlib

import pandas as pd
import matplotlib.pyplot as plt

df=pd.read_csv("market.csv")
df
```

```python
#Pie chart for average sales on product
avg_sales=df.groupby("Product")["Sales"].mean()
plt.figure(figsize=(8,8))
plt.pie(
    avg_sales.values,
    labels=avg_sales.index,
    autopct="%1.1f%%"
    )
plt.title("average sales on product")
plt.show()
```

```python
#Pie chart for total sales on products
total_sales=df.groupby("Product")["Sales"].sum()
plt.figure(figsize=(8,8))
plt.pie(
    total_sales,
    labels=total_sales.index,
    autopct="%1.1f%%",
    labeldistance=0.5,
    #startangle=45   #rotates whole pie chart
    )
plt.title("total sales on product")
plt.show()
```

```python
#total returns on product
total_returns=df.groupby("Product")["Returns"].sum()
plt.figure(figsize=(8,8))
plt.pie(
    total_returns,
    labels=total_returns.index,
    autopct="%1.1f%%"
    )
plt.title("Total returns on product")
plt.show()
```
