```python
#Conditional Statement
'''
1.if 
2.if else
3.if elif else
4.nested if
'''

age=int(input("enter the age"))
if age>0:
    if age>18:
        print("eligible for vote")
    elif age==18 or age>18:
        print("eligible for vote")
    else:
        print("not eligible for vote")
else:
    print("enter the correct age")
```
