```python
from scipy import stats
import numpy as np
age=np.array([24,35,46,51,65])
print(age)
print(age.ndim)   #shows the which dimention of data it is
```

```python
data=np.array([[12,14,15,19],[21,25,28,27],[32,36,31,38],[45,42,41,48]])
print(type(data))
print(data.ndim)
print(data)
```

```python
#Slicing
print(data[0:4,0:3])
print(data[:,0:2])
print(data[0:4:2,0:3:2])
```

```python
#calculations
print("Minimum value",data.min())
print("Max value",data.max())
print("Mean value",data.mean())
print("Standard deviation",data.std())
print("Variance",data.var())

#Calculation of particular rows and columns
print(data)
print(np.mean(data[1:2,1:3]))
print(np.median(data[0:3,0:2]))
print(stats.mode(data[0:4,0:3],keepdims=True))
data.shape
```
