```python
#Loops Statements
#1. for loop
print("----------for loop---------------")
l=[7,6,1,3,5]
for k in l:
    print(k)
#finding even and odd numbers
print("finding even and odd numbers")
list=[5,7,1,2,6,9,4,18]
for i in list:
    if i%2==0:
        print(i)
print("numbers 1 to 10")
for i in range(1,11):
    print(i)

print("print even numbers 1 to 20")
for i in range(1,21):
    if i%2==0:
        print(i)

print("find the sum of numbers 1 to 100")
sum=0
for i in range(1,101):
    sum=sum+i
print(sum)

print("count the vowels in the string")
str="MUBASHIR"
for i in str:
    if i in "AEIOU":
        print(i)

print("stars")
for i in range(1,6,+1):
    print("*"*i)
    




#2. While
print("------------While loop------------")
print("numbers 1 to 10")
i=1
while(i<=10):
    print(i)
    i=i+1

print("even and odd numbers")
list=[2,1,8,9,13,16,7]
even_list=[]
odd_list=[]
i=0
while(i<len(list)):
    if list[i]%2==0:
        even_list.append(list[i])
    else:
        odd_list.append(list[i])
    i=i+1
print(even_list)
print(odd_list)
```
