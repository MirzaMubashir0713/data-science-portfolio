```python
#Replacing the Null value with the mean or median value
#Fillna method is used to replace the null vlaue
mean=int(df1["age"].mean())
df1["age"].fillna(value=mean,inplace=True)   #inplace is using because it affect the actual data if we wont use its not affect
df1
```

```python
#Concatination of two files  (Rows are adding the axis is default 0)
df4=pd.concat([df1,df2],ignore_index=True)
df4
```

```python
#Concatination of two files columns (axis should be change to 1)
df5=pd.concat([df4,df3] , axis=1)
df5
```

```python
print("------------average age of male and female-------------------- ")
df5.groupby("Gender")["age"].mean()
```

```python
print("----------average height of male and female ------------")
df5.groupby("Gender")["height"].mean()
```

```python
print("---------average weight of male and female -----------------")
df5.groupby("Gender")["weight"].mean()
```
