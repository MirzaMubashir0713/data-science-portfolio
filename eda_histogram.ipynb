```python
#EDA (EXPLORATORY DATA ANALYST )
'''
we have two library for visualization 
     1. MATPLOTLIB
     --> it is a python used to create the visualization plot.
     2. SEABORN 
     --> it is advanced visualization library
     --> it is creates on top of matplotlib
'''

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df=pd.read_csv("market.csv")
df
```

```python
#HISTOGRAM graph
# its shows the frequency distribution of continuous numerical values
#from data print the sales frequecy distribution

# MATPLOTLIB VERSION FOR SALES
plt.figure(figsize=(8,5))
plt.hist(df["Sales"],bins=5)
plt.title("distribution on sales using matplotlib")
plt.xlabel("sales")
plt.ylabel("frequency")
plt.show()
```

```python
#SEABORN VERSION FOR SALES
plt.figure(figsize=(8,5))
sns.histplot(df["Sales"],bins=15,kde=True)  #Using Kde = True it shows the line
plt.title("distribution of sales using seaborn")
plt.xlabel("sales")
plt.ylabel("frequency")
plt.show
```

```python
# MATPLOTLIB VERSION OF INVENTORY
plt.figure(figsize=(8,5))
plt.hist(df["Inventory"],bins=10)
plt.title("distribution of inventory using matplotlib")
plt.xlabel("inventory")
plt.ylabel("frequency")
plt.show()
```

```python
#SEABORN VERSION FOR INVENTORY
plt.figure(figsize=(8,5))
sns.histplot(df["Inventory"],bins=15,kde=True)
plt.title("disttibution of inventory using seaborn")
plt.xlabel("inventory")
plt.ylabel("frequency")
plt.show()
```

```python
#MATPLOTLIB VERSION FOR RETURNS
plt.figure(figsize=(8,5))
plt.hist(df["Returns"],bins=8)
plt.title("distribution of return using matplotlib")
plt.xlabel("returns")
plt.ylabel("frequency")
plt.show()
```

```python
#SEABORN VERSION FOR RETURNS
plt.figure(figsize=(8,5))
sns.histplot(df["Returns"],bins=8,kde=True)
plt.title("distribution of returns using seaborn")
plt.xlabel("returns")
plt.ylabel("frequency")
plt.show()
```
