```python
#PROBABILITY 
#SCIPY is a library use to solve some statistics basic

'''Q1. The annual salaries of employees in a large company are approximately normally distributed with a mean of $50,000 
       and a standard deviation of $10,000.

a) What percent of people earn less than $60,000?

'''
from scipy.stats import norm
nd1=norm(50000,10000) #creating normal  distribution object
x1=nd1.cdf(60000) # it checks X<=60000 
print(x1)
x2=1-nd1.cdf(60000) #checks X>=60000  
#or x2=nd.sf(60000) checks X>=60000
print(x2)



```

```python
'''
Q2. A radar unit is used to measure speeds of cars on a highway. The speeds are normally distributed with a mean of 90 km/hr 
and a standard deviation of 10 km/hr.What is the probability that a car picked at random is travelling at more than 100 km/hr?

'''
nd2=norm(90,10)
y=1-nd2.cdf(100)
print(y)
```

```python

'''
For a certain type of mobiles, the length of time between charges of the battery is normally distributed with a mean of 10 hours and 
a standard deviation of 20 minutes.
John owns one of these mobiles and wants to know the probability that the length of time will be between 10 and 11 hours

'''

#mean=10 hr / 600 minutes
#STD = 20 minutes

nd3=norm(600,20)
x3=nd3.cdf(660)-nd3.cdf(600)
print(x3)
```
