```python
#python Operators 
'''
1. Arthimatic Operator
(+,-,*,/,//,**)
'''
print("----------------Arthimatic Operator-------------")
a=5
b=3
c1=a+b #add
c2=a-b #sub
c3=a*b #mul
c4=a/b #div returns float qutation
c5=a//b #floor returns int qutation
c6=a**b
print(c1)
print(c2)
print(c3)
print(c4)
print(c5)
print(c6)
'''
2. Assignment Operator
(=,+=,-=,*=,/=,//=,**=)
'''
print("----------------Assignment Operator---------------")
a=10
a+=20
print(a)
a-=20
print(a)
a*=5
print(a)
a/=5
print(a)
a=10
a//=10
print(a)
'''
3. Comparision Operator (returns boolean value either true or false)
(<,>,<=,>=,==,!=)
'''
print("------------comaprision operators-------------")
x=5
y=10
z1=x<y
z2=x>y
z3=x<=y
z4=x>=y
z5=x==y
z6=x!=y
print(z1)
print(z2)
print(z3)
print(z4)
print(z5)
print(z6)
'''
4. Logical Operator
(and , or , not)
and- if all true then only returns true otherwise false
or- if all false then only returns false otherwise false
not- true-->false or false-->true
'''
print("----------Logical operators--------------------")
num1=20
num2=40
num3=20
print("and Operator")
print(num1>num2 and num1<=num3)
print(num1==num3 and num1<num2)
print("and Operator")
print(num1>num2 or num1<=num3)
print(num1==num3 or num1<num2)
print("not Operator")
changes=(num1==num3 or num1<num2)
print(changes)
print(not changes)
'''
5. Bitwise Operator 
(&,|,!,<<,>>,^,~)
it first convert the code in binary numbers then apply the operations on that
'''
print("-------------Bitwise Operators-------------------")
a=8
b=12
c1=a&b
c2=a|b
print(c1)
print(c2)
print(~b)  #bitwise Xor flip all the binary values 
'''
6. Membership Operator (return boolean values)
check specific value is present or not 
(in , not in)
'''
print("------------Membership Operator----------------")
M=["Mubashir", "Aman", "Zahir","Atif"]
print("mubashir" in M)
print("Aman" in M )
print("Zahir" not in M)
print("atif" not in M )


'''
7. Identity Operator
check the memory addres is same or not
---> PYTHON STORES SAME MEMORY ADDRESS TILL 256 VALUES ONLY NUMERICAL VALUES
(is , is not)

'''
print("------------Identity Operator-----------------")
K=20
L=20
print(id(K))
print(id(L))
print(K is L) 
print(K is not L)
```
